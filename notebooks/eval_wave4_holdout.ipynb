```python
from config import setting_path
from ingest.sqlite import connect
```

```python
persona_db = setting_path("PERSONA_DB")
```

```python
with connect(persona_db) as conn:
    question_rows = conn.execute(
        """
        SELECT q.item_id, q.question_id, q.sub_id, q.question_text, q.item_text, q.options_json, q.question_type
        FROM questions AS q
        WHERE LOWER(q.question_type) IN ("mc", "matrix")
        AND q.is_overlap = 1
        ORDER BY q.item_id
        """
    ).fetchall()
```

```python
distinct_questions = len(set([row[1] for row in question_rows]))
print(f"Distinct questions: {distinct_questions}")
```

```python
matrix_questions = pd.DataFrame([row for row in question_rows if row[-1].lower() == "matrix"], 
        columns=["item_id", "question_id", "sub_id", "question_text", "item_text", "options_json", "question_type"]).drop(columns=["question_type"]).reset_index(drop=True)

        
matrix_questions.head()
# print(f"Matrix questions: {len(matrix_questions)}")
```

```python
display(matrix_questions.sort_values("item_id"))
```

```python
print(f"Total items: {len(matrix_questions)}")
print(f"Distinct questions: {len(set(matrix_questions['question_id']))}")
```

```python
print(f"Total items: {len(MC_questions)}")
print(f"Distinct questions: {len(set(MC_questions['question_id']))}")
```
